```python
from torchgeo.datasets import LandCoverAI
import numpy as np

raw_dataset = LandCoverAI(root="data", split="train", download=True)

print("Number of images:", len(raw_dataset))
sample = raw_dataset[0]
print("Image shape:", sample["image"].shape)
print("Mask shape:", sample["mask"].shape)
print("Classes present in this mask:", sample["mask"].unique())
```

```python
def build_pairs(raw_dataset, n_samples=200):
    pairs = []
    for i in range(min(n_samples, len(raw_dataset))):
        sample = raw_dataset[i]
        img = sample["image"].permute(1, 2, 0).numpy().astype(np.float32) / 255.0  # CHW -> HWC, normalize
        mask = sample["mask"].numpy().astype(np.int64)
        pairs.append((img, mask))
    return pairs

train_pairs = build_pairs(raw_dataset, n_samples=200)
print("Built", len(train_pairs), "real training pairs")
print("Example image shape:", train_pairs[0][0].shape, "mask shape:", train_pairs[0][1].shape)
```

```python
import json

points_values = [5, 20, 50]
results = []

for p in points_values:
    print(f"\n=== Training with {p} points per class (real data) ===")
    _, history = train_one_config(
        train_pairs_split, val_pairs, n_classes=n_classes, points_per_class=p,
        epochs=8, pretrained=True, device="cuda",
    )
    best_miou = max(h["val_miou"] for h in history)
    results.append({"points_per_class": p, "best_miou": best_miou})

print("\n================ SUMMARY (real data) ================")
for r in results:
    print(f"points_per_class={r['points_per_class']:>3}  ->  best mIoU={r['best_miou']:.4f}")

with open("experiment_results_real.json", "w") as f:
    json.dump(results, f, indent=2)
```
